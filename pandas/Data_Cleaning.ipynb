```python
import pandas as pd
df = pd.read_csv('AB_NYC_2019.csv')

#https://www.kaggle.com/datasets/dgomonov/new-york-city-airbnb-open-data?resource=download

print(df.head)
```

```python
#check for missing values

df.isna().sum()
```

```python
df['name'] = df['name'].fillna('Unknown')
df['host_name'] = df['host_name'].fillna('Unknown')
df['last_review'] = df['last_review'].fillna('Unknown')
df['reviews_per_month'] = (df['reviews_per_month'].fillna(0))

df.isna().sum()
```

```python
#check for duplicate values
df.duplicated().sum()

df.drop_duplicates()
```

```python
#convert data type
print(df.dtypes)

df['last_review'] = pd.to_datetime(
    df['last_review'],
    errors='coerce'
)
print(df.dtypes)
```

```python
#group by borough
print(df.groupby('neighbourhood_group')['price'].mean().reset_index())
```

```python
#10 most expensive listings
expensive_listings = df.sort_values(
    by='price',
    ascending=False
)

print(expensive_listings[['name', 'neighbourhood_group', 'price']].head(10))
```
